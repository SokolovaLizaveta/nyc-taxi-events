```python
# посмотрим мероприятия, прошедшие в 2025 году и их количество
SETLIST_VENUE_ID = venue_setlist_data["venue"][1]["id"]
url_setlists = "https://api.setlist.fm/rest/1.0/search/setlists"

params_setlists = {
    "venueId": SETLIST_VENUE_ID,
    "year": "2025",
    "p": 1
}
response_setlists = requests.get(
    url_setlists,
    params=params_setlists,
    headers=headers_setlist,
    timeout=30
)
response_setlists.status_code
```

```python
setlists_data = response_setlists.json()
setlists_data.keys()
```

```python
# Setlist.fm нашёл 179 записей для Madison Square Garden за 2025 год
# API возвращает по 20 записей на страницу, поэтому необходимо собрать все страницы

print('Всего записей за 2025 год:', setlists_data["total"])
print('Записей на одной странице:', len(setlists_data["setlist"]))
```

```python
import math
import time

total = setlists_data["total"]
items_per_page = setlists_data["itemsPerPage"]

total_pages = math.ceil(total / items_per_page)

# Первая страница у нас уже есть
all_setlists = setlists_data["setlist"].copy()

# Поэтому начинаем со второй страницы
for page in range(2, total_pages + 1):
    
    time.sleep(3.2)

    params_setlists = {
        "venueId": SETLIST_VENUE_ID,
        "year": "2025",
        "p": page
    }

    response = requests.get(
        url_setlists,
        params=params_setlists,
        headers=headers_setlist,
        timeout=30
    )

    if response.status_code != 200:
        print(f"Ошибка на странице {page}: {response.status_code}")
        break

    page_data = response.json()
    all_setlists.extend(page_data["setlist"])
```

```python
# Собрали все 179 записей с 9 страниц
len(all_setlists)
```

```python
# ЧАСТЬ 7: TAXI ZONES КОНЦЕРТНЫХ ПЛОЩАДОК
# Координаты площадок получены вручную из Google Maps.
# Для каждой площадки координаты используются для определения Taxi Zone.
venues = pd.DataFrame({
    "venue": [
        "Madison Square Garden",
        "Isaac Stern Auditorium",
        "Radio City Music Hall",
        "Beacon Theatre",
        "Webster Hall"
    ],
    "latitude": [
        40.750591520873265,
        40.76514620273492,
        40.75999993740611,
        40.78050757471799,
        40.731705852947016
    ],
    "longitude": [
        -73.9934420610122,
        -73.97992017577518,
        -73.9799781831889,
        -73.9811761297521,
        -73.98922140568371
    ]
})

venues
```

```python
# Проверка на правильность заполнения названий площадок
set(venues["venue"]) == set(all_event_days["venue"].unique())
```

```python
import geopandas as gpd

taxi_zones = gpd.read_file(
    "../data/raw/taxi_zones/taxi_zones.shp"
)
taxi_zones.crs
```

```python
venues_gdf = gpd.GeoDataFrame(
    venues,
    geometry=gpd.points_from_xy(
        venues["longitude"],
        venues["latitude"]
    ),
    crs="EPSG:4326"
)
venues_gdf
```

```python
venues_gdf = venues_gdf.to_crs(taxi_zones.crs)
venues_gdf.crs == taxi_zones.crs
```

```python
# При выборе площадок я учитывала их географическое расположение
# Все выбранные площадки находятся в разных Taxi Zones Манхэттена,
# что позволяет анализировать локальный спрос в нескольких частях города.
# Isaac Stern Auditorium и Radio City Music Hall находятся относительно близко,
# но относятся к разным Taxi Zone.
# Isaac Stern Auditorium оставляю для разнообразия музыкального профиля:
# площадка ориентирована в том числе на классическую музыку.

venues_with_zones = gpd.sjoin(
    venues_gdf,
    taxi_zones[
        ["LocationID", "zone", "borough", "geometry"]
    ],
    how="left",
    predicate="within"
)
venues_with_zones 
```

```python
all_event_days = all_event_days.merge(
    venues_with_zones[
        ["venue", "LocationID", "zone"]
    ],
    on="venue",
    how="left"
)

all_event_days
```
